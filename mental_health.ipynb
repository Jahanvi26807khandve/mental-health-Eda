```python
latest_year = df['Year'].max()
df_latest_year = df[df['Year'] == latest_year]

plt.figure(figsize=(14, 8))
sns.barplot(
    data=df_latest_year.sort_values(by="Depressive disorders (share of population) - Sex: Both - Age: Age-standardized", ascending=False).head(20),
    x="Depressive disorders (share of population) - Sex: Both - Age: Age-standardized",
    y="Entity",
    palette='viridis'
)

plt.title(f"Top 20 Entities by Depressive Disorders Prevalence in {latest_year}")
plt.xlabel("Depressive Disorders Prevalence (share of population)")
plt.ylabel("Entity")
plt.tight_layout()
plt.show()
```
